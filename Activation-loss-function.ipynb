```python
def sigmoid(x):
    return 1 /  (1 + np.exp(-x)) 

def sigmoid_derivative(z):
    s = sigmoid(z) 
    return s * (1-s) 

z = np.array([-2,-1,0,1,2]) 

output = sigmoid(z)
print(output)
```
