# Types of gradient descent 
1. Batch Gradient descent - weight update depends on epoch
2. Stochastic Gradient descent
   epoch = 5
   for i in range(5):
    do shuffle to remove bisasness
     for j in range(X.shape[0]):
       select 1 row
       -> predict or forward propagation
       -> loss calculate
       -> weight update 
4. Mini Batch Gradient Descent

```python
import torch 
import torch.nn as nn 
import torch.optim as optim 
from torch.utils.data import DataLoader, TensorDataset 
import matplotlib.pyplot as plt 
```

```python
# -------------------------------------------------
# Create Synthetic Dataset (50 samples, 2 features)
# -------------------------------------------------

torch.manual_seed(42)

X = torch.randn(50, 2)

true_w = torch.tensor([[2.0], [-3.0]])
true_b = 1.0

y = torch.sigmoid(X @ true_w + true_b)
y = (y > 0.5).float()   # Convert to binary 0/1

dataset = TensorDataset(X, y)
```

```python
# -------------------------------------------------
# Define Simple Model
# -------------------------------------------------

class SimpleModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(2, 1)

    def forward(self, x):
        return torch.sigmoid(self.linear(x))
```
